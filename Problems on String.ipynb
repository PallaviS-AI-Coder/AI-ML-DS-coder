Write a Python program to count vowels and consonants in a string.
    
* Input Programming
    
* Expected Output Vowels = 3, Consonants = 8

```python
print("2. Count vowels and consonants")
s = "Programming"
vowels = 0
consonants = 0
for ch in s:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1
print(f"Input: {s}")
print(f"Vowels = {vowels}, Consonants = {consonants}")
print()

```


3. Check whether a string is a palindrome or not
      * input madam

```python

s = "madam"
if s == s[::-1]:
    print(f"Input: {s}")
    print("The string is a palindrome.")
else:
    print(f"Input: {s}")
    print("The string is not a palindrome.")
print()
```

10 Write a Python program to find the frequency of each character in a string.
* Input aabcc
* Expected Output a: 2, b: 1, c: 2

```python

s = "aabcc"
freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1
print(s)
print(", ".join(f"{k}: {v}" for k, v in freq.items()))
print()

```

```python

l = "aabcc"
# create dictionary
frequency = {}

for char in l:
    if char in frequency:
        frequency[char] = frequency[char] + 1   # if same char comes it will add +1
    else:
        frequency[char] = 1
print( frequency)      
```

16 . Write a Python program to remove duplicate characters from a string while preserving order.
* Input programming
* Expected Output Unique characters = progamin

```python

q = 'programming'

uniq = ''

for char in q:
  if char not in uniq:
    uniq = uniq + char

print('Unique characters = ', uniq)
```

16 Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a string.
* Input LihaTech@2026
* Expected Output Uppercase = 2, Lowercase = 6, Digits = 4, Special = 1

```python
r = 'LihaTech@2026'
uppercase = 0
lowercase = 0
digits = 0
special = 0
space = 0

for char in r:
    if char.isupper():
        uppercase = uppercase + 1
    elif char.islower():
        lowercase = lowercase + 1
    elif char.isdigit():
        digits = digits + 1
    elif char == ' ' :
        space = space + 1
    else:
        special = special + 1

print("Uppercase =", uppercase, ", Lowercase =", lowercase, ", Digits =", digits, ", Special =", special, ", Space =", space)
```
